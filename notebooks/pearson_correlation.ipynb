```python
import pandas as pd
import numpy as np

file_path = '../data/boston_weather_data.csv'
df = pd.read_csv(file_path)

df['target_tomorrow_tavg'] = df['tavg'].shift(-1) 
df_clean = df.dropna()

candidate_features = ['tavg', 'tmin', 'tmax', 'prcp', 'wdir', 'wspd', 'pres']
available_features = [col for col in candidate_features if col in df_clean.columns]

correlations = df_clean[available_features].corrwith(df_clean['target_tomorrow_tavg'], method = 'pearson')
print("the correlation of each feature with tmr's temp:")
print(correlations)
print(f"# of rows of the dataset: {len(df_clean)}\n")


candidate_features2 = ['tavg', 'tmin', 'tmax', 'prcp', 'wspd', 'pres']
df_clean2 = df.dropna(subset = candidate_features2 + ['target_tomorrow_tavg'])

correlations2 = df_clean2[candidate_features2].corrwith(df_clean2['target_tomorrow_tavg'], method = 'pearson')
print("the correlation of each feature(without wdir) with tmr's temp:")
print(correlations2)
print(f"# of rows of the dataset: {len(df_clean2)}\n")

feature_corr_matrix = df_clean[candidate_features].corr(method='pearson')

print("the correlation between each feature:")
print(feature_corr_matrix[['tavg', 'tmin', 'tmax', 'prcp', 'wdir', 'wspd', 'pres']])
```
